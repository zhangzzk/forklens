## Shape measurement

```python
import csst_sim as cs
from network import ForkCNN

import torch
from torch import nn
from torch import optim
from torch.utils.data import Dataset, DataLoader, SubsetRandomSampler
```

```python
class ShapeDataset(Dataset):
    def __init__(self, dataset):
        
        self.classes_frame = dataset

    def __len__(self):
        return self.classes_frame.shape[0]
    
    def __set_pars(self, idx):
        
        param_gals = {}
        param_gals["z"] = 1
        param_gals["e1"] = self.classes_frame[idx,0]
        param_gals["e2"] = self.classes_frame[idx,1]
        param_gals["hlr_disk"] = self.classes_frame[idx,2]
        param_gals["mag_i"] = self.classes_frame[idx,3]
        self.gal_pars = param_gals

        param_psfs = {}
        param_psfs['randint'] = self.classes_frame[idx,4]
        self.psf_pars = param_psfs

    def __getitem__(self, idx):
        
        self.__set_pars(idx)

        gal_image, clean_gal, psf_image, label_ = cs.get_gal(
            self.gal_pars,
            self.psf_pars,
            shear=None,
            stamp_size=128,
        )
        
        label = np.array([label_[0], label_[1], label_[2]/1.2, label_[3]/25])
        SNR = snr.true_snr(gal_image,clean_gal)
        
        return {'gal_image': gal_image[None], 'psf_image': psf_image[None], 'label': label, 'snr':SNR, 'id': idx}

def predictFuncs(MODEL, DATALOADER, CRITERION=nn.MSELoss()):
    MODEL.eval()

    losses=[]
    for i, batch in enumerate(DATALOADER):
        inputs1, inputs2 = batch['gal_image'].float().to(device), \
                                   batch['psf_image'].float().to(device)
        outputs = MODEL.forward(inputs1, inputs2)
        labels_true_batch = batch['label'].float().view(-1,nFeatures).to(device)
        SNR = batch['snr'].float()#.to(device)
        loss = CRITERION(outputs, labels_true_batch)
        losses.append(loss.item())
        if i == 0:
            ids = batch['id'].numpy()
            labels = outputs.detach().cpu().numpy()
            labels_true = labels_true_batch.cpu()
            snr_true = SNR.numpy()
        else:
            ids = np.concatenate((ids, batch['id'].numpy()))
            labels = np.vstack((labels, outputs.detach().cpu().numpy()))
            labels_true = np.vstack((labels_true, labels_true_batch.cpu()))    
            snr_true = np.concatenate((snr_true, SNR.numpy()))
                
    combined_pred = np.column_stack((ids, labels))
    combined_true = np.column_stack((ids, labels_true))
    # combined_snr = np.column_stack((ids, snr_true))

    epoch_loss = np.sqrt(sum(losses) / len(losses))
    return combined_pred, combined_true, snr_true, epoch_loss
```

```python
BATCH_SIZE = 200
nFeatures = 4
GPU_num = 1

# Device Options
if torch.cuda.is_available():
    DEVICE = "cuda:0" # "cuda" or "cpu"
    NR_DEVICES = torch.cuda.device_count()
else:
    DEVICE = "cpu"
    NR_DEVICES = torch.get_num_threads()    
NR_WORKERS = 10 * NR_DEVICES

device = torch.device(DEVICE)
model = ForkCNN(BATCH_SIZE,GPU_num,nFeatures)
model = nn.DataParallel(model, device_ids=range(GPU_num))
model.to(device)
model_file = "./model/test_200,000_all"

model.load_state_dict(torch.load(model_file), strict=False)
```

```python
with fits.open('./things/disk_gal_catalog_cut_25_1.2_snr.fits') as f:
    data = f[0].data
    idx = np.where((data[:,2]>0)&(data[:,0]>0.1))[0]
    mag_cat = data[idx,1]
    size_cat = data[idx,0]
    
# plt.figure(figsize=(5,5))
# plt.scatter(mag_cat[::10], size_cat[::10], s=1, alpha=0.2)
# print(mag_cat.shape[0])

def dataframe(num, seed):
    
    num = int(num/2)
    rng1 = np.random.RandomState(seed)
    Gal_Hlr   = size_cat[rng1.randint(0,mag_cat.shape[0],size=num)]
    Gal_Hlr   = np.concatenate((Gal_Hlr, Gal_Hlr), axis=0)
    Gal_Mag   = mag_cat[rng1.randint(0,mag_cat.shape[0],size=num)]
    Gal_Mag   = np.concatenate((Gal_Mag, Gal_Mag), axis=0)
    
    Gal_Theta = rng1.random(num*2)*2*np.pi
    Gal_ELL = np.sqrt(rng1.random(num*2))*0.999
    e1 = Gal_ELL*np.sin(2*Gal_Theta)
    e2 = Gal_ELL*np.cos(2*Gal_Theta)

    rng2 = np.random.RandomState(seed+1)
    PSF_randint = rng2.randint(0,10000,size=num)
    PSF_randint = np.concatenate((PSF_randint, PSF_randint), axis=0)

    Dataframe        = np.zeros((num*2, 5))
    Dataframe[:,0]   = e1
    Dataframe[:,1]   = e2
    Dataframe[:,2]   = Gal_Hlr
    Dataframe[:,3]   = Gal_Mag
    Dataframe[:,4]   = PSF_randint
    
    return Dataframe

nSims = 1000
Dataframe       = dataframe(nSims, 123)

test_ds = ShapeDataset(Dataframe)
test_dl = DataLoader(test_ds, 
                      batch_size=BATCH_SIZE, 
                      num_workers=NR_WORKERS)

# shape measurement with CNN model
pred, true, SNR, loss = predictFuncs(model, test_dl)
print('LOSS:', loss)
```

```python
title = ['gal_e1','gal_e2','gal_hlr','gal_magi']

# plotting measurement residuals of galaxies
idx = np.where(SNR>10)[0]

plt.figure(figsize=(20,5))
for i in range(nFeatures): 
    plt.subplot(1,4,i+1)
    plt.title(title[i],fontsize=18)
    plt.scatter(true[idx,i+1], pred[idx,i+1],c=SNR[idx],s=8,alpha=0.5)
    plt.plot(true[idx,i+1], true[idx,i+1], 'k-', linewidth=2)
    plt.tick_params(axis='both',which='major',labelsize=16)
    
plt.colorbar()
plt.tight_layout()
```

```python
def ellipse(e1,e2,size):
    
    e = np.sqrt(e1**2+e2**2)
    if e1 < 0:
        phi = np.arctan(e2/e1)/2 +np.pi/2
    else:
        phi = np.arctan(e2/e1)/2
    
    axis_ratio = (1-e) / (1+e)
    a = size/np.sqrt(axis_ratio**2+1)  
    b = a * axis_ratio      
    t_rot = phi

    t = np.linspace(0, 2*np.pi, 100)
    Ell = np.array([a*np.cos(t) , b*np.sin(t)])  
    R_rot = np.array([[np.cos(t_rot) , -np.sin(t_rot)],[np.sin(t_rot) , np.cos(t_rot)]])  

    Ell_rot = np.zeros((2,Ell.shape[1]))
    for i in range(Ell.shape[1]):
        Ell_rot[:,i] = np.dot(R_rot,Ell[:,i])
        
    return Ell_rot

x = np.where((SNR[:]<50)&(SNR[:]>40))

rng = np.random.RandomState(198)
y = rng.randint(len(x[0]),size=4)
plt.figure(figsize=(16,8))
for i,idx in enumerate(y):
    
    out = test_ds.__getitem__(x[0][idx])
    ell_true = ellipse(out['label'][0],out['label'][1],out['label'][2]/0.074*12)
    ell = ellipse(pred[x[0][idx],1],pred[x[0][idx],2],pred[x[0][idx],3]/0.074*12)
    
    plt.subplot(2,4,i+1)
    plt.imshow(out['gal_image'][0],cmap='gray')
    
    plt.plot(ell_true[0,:]+63.5, ell_true[1,:]+63.5,'white',linewidth=3)
    plt.plot(ell[0,:]+63.5, ell[1,:]+63.5,'purple',linewidth=3)
    
    plt.text(10,20,round(SNR[x[0][idx]],ndigits=2),fontsize=24,color='white')
    plt.xticks([])
    plt.yticks([])

plt.subplots_adjust(wspace=0., hspace=-0.3)
```

```python
file = "./things/nn_typeII/TypeII_dataset_%scase_%sreal_0.1shear_0.fits"%(case_num,real_num)
dataset = fits.getdata(file)

# Get proper format
inputs = dataset.reshape(case_num,real_num,-1).transpose(1,2,0)

g = np.mean(inputs[:,5:7,:], axis=0) # measred g
labels = inputs[0,9:11,:]            # true g
```

```python
plt.figure(figsize=(6,6))

plt.scatter(labels[0,:], labels[1,:], marker='x', label='input')
plt.scatter(g[0,:], g[1,:], label='CNN measurement')

plt.xlim(-0.1,0.1)
plt.ylim(-0.1,0.1)

plt.xlabel('g1')
plt.ylabel('g2')
plt.legend()

ax = plt.gca()
ax.set_aspect('equal', adjustable='box')
plt.draw()
```

```python
normalization = 0.1

ten_g1 = tenbilac.com.Tenbilac('../ten_wd/cali_g1/')
ten_g2 = tenbilac.com.Tenbilac('../ten_wd/cali_g2/')
g1_cali_ = ten_g1.predict(inputs[:,5:9,:])*normalization
g2_cali_ = ten_g2.predict(inputs[:,5:9,:])*normalization

g1_cali = np.mean(g1_cali_, axis=0)
g2_cali = np.mean(g2_cali_, axis=0)

ten_w_g1 = tenbilac.com.Tenbilac('../ten_wd/weight_g1/')
ten_w_g2 = tenbilac.com.Tenbilac('../ten_wd/weight_g2/')
w_g1 = ten_w_g1.predict(inputs[:,5:9,:])
w_g2 = ten_w_g2.predict(inputs[:,5:9,:])

g1_weigh = np.mean(g1_cali_*w_g1, axis=0)/np.mean(w_g1, axis=0)
g2_weigh = np.mean(g2_cali_*w_g2, axis=0)/np.mean(w_g2, axis=0)
```

```python
plt.figure(figsize=(6,6))

plt.scatter(labels[0,:], labels[1,:], marker='x', label='input')
plt.scatter(g1_cali[0,:], g2_cali[0,:], label='calibrated')
plt.scatter(g1_weigh[0,:], g2_weigh[0,:], label='weighted average')

plt.xlim(-0.1,0.1)
plt.ylim(-0.1,0.1)

plt.xlabel('g1')
plt.ylabel('g2')
plt.legend()

ax = plt.gca()
ax.set_aspect('equal', adjustable='box')
plt.draw()
```
